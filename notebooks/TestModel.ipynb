```python
import pandas as pd
import joblib
import math

model = joblib.load("../models/best_gbr.pkl")
encoder = joblib.load("../models/encoders.pkl")
scaler = joblib.load("../models/scaler.pkl")
```

```python
company = input("Company: ")
typename = input("Laptop Type: ")

inches = float(input("Screen Size (Inches): "))
ram = float(input("RAM (GB): "))

opsys = input("Operating System: ")
weight = float(input("Weight (kg): "))

touchscreen = int(input("Touchscreen? (0 = No, 1 = Yes): "))

resolution = input("Resolution (e.g. 1920x1080): ")
x_res, y_res = resolution.split("x")

x_res = float(x_res)
y_res = float(y_res)

clock_speed = float(input("Clock Speed (GHz): "))

processor = input("Processor Type: ")
gpu_brand = input("GPU Brand: ")

ssd = float(input("SSD (GB): "))
hdd = float(input("HDD (GB): "))
flash_storage = float(input("Flash Storage (GB): "))
hybrid = float(input("Hybrid Storage (GB): "))
```

```python
ips = int(input("IPS? (0 = No, 1 = Yes): "))
retina = int(input("Retina Display? (0 = No, 1 = Yes): "))
```

```python
ppi = math.sqrt(
    x_res**2 + y_res**2
) / inches
```

```python
data = pd.DataFrame([{
    "Company": company,
    "TypeName": typename,
    "Inches": inches,
    "Ram": ram,
    "OpSys": opsys,
    "Weight": weight,
    "IS_Touchscreen": touchscreen,
    "IPS": ips,
    "Retina Display": retina,
    "X_res": x_res,
    "Y_res": y_res,
    "PPI": ppi,
    "Clock_Speed_GHz": clock_speed,
    "Processor_Type": processor,
    "Gpu_Brand": gpu_brand,
    "SSD": ssd,
    "HDD": hdd,
    "Flash_Storage": flash_storage,
    "Hybrid": hybrid
}])
```

```python
categorical_cols = [
    "Company",
    "TypeName",
    "OpSys",
    "Processor_Type",
    "Gpu_Brand"
]

numerical_cols = [
    "Inches",
    "Ram",
    "Weight",
    "X_res",
    "Y_res",
    "PPI",
    "Clock_Speed_GHz",
    "SSD",
    "HDD",
    "Flash_Storage",
    "Hybrid"
]

encoded = encoder.transform(data[categorical_cols])

encoded_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(categorical_cols)
)

scaled = scaler.transform(data[numerical_cols])

scaled_df = pd.DataFrame(
    scaled,
    columns=numerical_cols
)

binary_df = data[
    ["IS_Touchscreen", "IPS", "Retina Display"]
]

final_data = pd.concat(
    [scaled_df, encoded_df, binary_df],
    axis=1
)

final_data = final_data[
    model.feature_names_in_
]
```

```python
prediction = model.predict(final_data)[0]

print(f"Predicted Laptop Price: $ {prediction:,.0f}")
```
