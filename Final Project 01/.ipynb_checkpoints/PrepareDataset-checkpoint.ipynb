
## About the data
This dataset contains detailed specifications of 3,976 laptops from various brands. The data includes a variety of features like brand, processor details, RAM, storage options, GPU information, display type, price, and battery life.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('laptopData(Target column Price).csv')
df.head()
```

```python
df.info()
```

```python
df['Weight'] = (df['Weight'].str[:-2])
df['Weight'] =pd.to_numeric(df['Weight'], errors='coerce')
df['Ram'] = (df['Ram'].str[:-2])
df['Ram'] = pd.to_numeric(df['Ram'], errors='coerce')
```

```python
df.head()
```

```python
df.shape
```

```python
df.drop('Unnamed: 0',axis=1,inplace=True)
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
# the Cpu ,Ram,Memory,Storage,Display,Graphics,Operating System columns are categorical columns and we need to replace the null values with the mode.
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='most_frequent')
for col in df.select_dtypes(include=['object']).columns:
    df[[col]] = imputer.fit_transform(df[[col]])
```

```python
df.select_dtypes(include=['object']).columns
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df.Inches.skew()
# right skewed , so we will replace the null values with median
imputer = SimpleImputer(strategy='median')
df[['Inches']] = imputer.fit_transform(df[['Inches']])
```

```python
df.Weight.skew()
# right skewed , so we will replace the null values with median
imputer = SimpleImputer(strategy='median')
df[['Weight']] = imputer.fit_transform(df[['Weight']])
```

```python
df.Ram.skew()
# right skewed , so we will replace the null values with median
imputer = SimpleImputer(strategy='median')
df[['Ram']] = imputer.fit_transform(df[['Ram']])
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df.dropna(inplace=True)
df.reset_index(drop=True, inplace=True)
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
df.reset_index(drop=True, inplace=True)
```

```python
# feature selection
df['IS_Touchscreen'] = df['ScreenResolution'].apply(lambda x: 1 if 'Touchscreen' in x else 0)
df['IPS'] = df['ScreenResolution'].apply(lambda x: 1 if 'IPS Panel' in x else 0)
df['Retina Display'] = df['ScreenResolution'].apply(lambda x: 1 if 'Retina Display' in x else 0)
res_extracted = df['ScreenResolution'].str.extract(r'(\d+)x(\d+)')
df['X_res'] = res_extracted[0].astype(int)
df['Y_res'] = res_extracted[1].astype(int)
df['PPI']=(((df['X_res']**2) + (df['Y_res']**2))**0.5 / df['Inches']).astype(float)
```

```python
df.drop('ScreenResolution',axis=1,inplace=True)
```

```python
df.head()
```

```python
df.Cpu.value_counts()
```

```python
df['Clock_Speed_GHz'] = (
    df['Cpu']
    .str.extract(r'(\d+(?:\.\d+)?)\s*GHz', expand=False)
    .astype(float)
)
```

```python
def extract_processor_type(cpu_string):
    if 'Intel Core i7' in cpu_string:
        return 'Intel Core i7'
    elif 'Intel Core i5' in cpu_string:
        return 'Intel Core i5'
    elif 'Intel Core i3' in cpu_string:
        return 'Intel Core i3'
    elif 'Intel' in cpu_string:
        return 'Other Intel Processor'
    elif 'AMD' in cpu_string:
        return 'AMD Processor'
    else:
        return 'Other'

df['Processor_Type'] = df['Cpu'].apply(extract_processor_type)
```

```python
df[df['Processor_Type']=='other']
df.head()
```

```python
df.drop('Cpu',axis=1,inplace=True)
```

```python
df.Gpu.value_counts()
```

```python
df['Gpu_Brand'] = df['Gpu'].apply(lambda x: x.split()[0])
```

```python
df.drop('Gpu',axis=1,inplace=True)
```

```python
df.head()
```

```python
df.Memory.value_counts()
```

```python
df=df[df['Memory']!='?']
df.reset_index(drop=True,inplace=True)
```

```python
df.Memory.value_counts()
```

```python
df['Memory'] = df['Memory'].astype(str).replace('\.0', '', regex=True)
df['Memory'] = df['Memory'].str.replace('GB', '')
df['Memory'] = df['Memory'].str.replace('TB', '000')
```

```python
# Extract SSD
df['SSD'] = df['Memory'].str.extract(r'(\d+(?:\.\d+)?) SSD')[0].fillna(0).astype(int)

# Extract HDD
df['HDD'] = df['Memory'].str.extract(r'(\d+(?:\.\d+)?) HDD')[0].fillna(0).astype(int)

# Extract Flash Storage
df['Flash_Storage'] = df['Memory'].str.extract(r'(\d+(?:\.\d+)?) Flash Storage')[0].fillna(0).astype(int)

# Extract Hybrid
df['Hybrid'] = df['Memory'].str.extract(r'(\d+(?:\.\d+)?) Hybrid')[0].fillna(0).astype(int)
```

## Encoding

```python
df.shape
```

```python
for col in df.select_dtypes(include='object'):
    print(f"{col} : {df[col].nunique()}") 
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
for col in df.select_dtypes(include='object'):
    one_hot=OneHotEncoder(sparse_output=False,drop='first')
    arr=one_hot.fit_transform(df[[col]])
    for index, n_col in enumerate(one_hot.get_feature_names_out()):
        df[n_col]=arr[:,index]
    df.drop(col,axis=1,inplace=True)
```

```python
df.select_dtypes(include='object').columns
df.head()
```

```python
## Splitting
```

```python
x=df.drop('Price',axis=1)
y=df['Price']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.15,random_state=12)
```

```python
feature_names = X_train.columns
```

## Scaling

```python
cols=['Inches', 'Ram',  'Weight',
    'X_res','Y_res', 'PPI',
    'Clock_Speed_GHz', 'SSD', 'HDD',
       'Flash_Storage', 'Hybrid']
```

```python
from sklearn.preprocessing import MinMaxScaler,RobustScaler,StandardScaler
scaler=RobustScaler()
X_train[cols]=scaler.fit_transform(X_train[cols])
X_test[cols]=scaler.transform(X_test[cols])
```

```python
X_train_df = pd.DataFrame(
    X_train,
    columns=feature_names
).reset_index(drop=True)

X_test_df = pd.DataFrame(
    X_test,
    columns=feature_names
).reset_index(drop=True)

y_train_df = pd.DataFrame(y_train).reset_index(drop=True)
y_test_df = pd.DataFrame(y_test).reset_index(drop=True)


train_data = pd.concat([X_train_df, y_train_df], axis=1)
test_data = pd.concat([X_test_df, y_test_df], axis=1)


train_data.to_csv('cleaned_train.csv', index=False)
test_data.to_csv('cleaned_test.csv', index=False)
```

```python
import joblib
joblib.dump(scaler, "scaler.pkl")
joblib.dump(one_hot, "encoder.pkl")
```
